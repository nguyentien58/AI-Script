### ***TENSOR***

```python
import torch
import numpy as np
```

```python
# Scarlar
a = torch.tensor(5)
# Vector
b = torch.tensor([1,2,3,4,5,6,8])
# Matrix
c = torch.tensor([
    [2,3,4],
    [4,5,6]])
# Tensor 3D
d = torch.zeros(2,3,4)

#Print a
print(f"Scarlar: {a}")
print(f"Scarlar shape: {a.shape}")
print(f"Scarlar type: {a.dtype}")
print('\n')
#Print b
print(f"Vector: {b}")
print(f"Vector shape: {b.shape}")
print(f"Vector type: {b.dtype}")
print('\n')
#Print c
print(f"Matrix: {c}")
print(f"Matrix shape: {c.shape}")
print(f"Matrix type: {c.dtype}")
print('\n')
#Print d
print(f"Tensor 3D {d}")
print(f"Tensor 3D shape: {d.shape}")
print(f"Tensor 3D type: {d.dtype}")

```

```python
#We can app numpy array to tensor
data = [[5,8],[4,5]]
np_array = np.array(data)
x = torch.from_numpy(np_array)
print(f"x = {x}")
print(f"x shape = {x.shape}")
```
